```python
import os 
import librosa
from joblib import Parallel, delayed
import pandas as pd
import numpy as np
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler
```

```python
# Unzip the GTZAN dataset downloaded from kaggle and place it in the same location as this notebook.
# The data set has a top level folder for "Data"
# All the audio files (.wav) are located in genres_original with subfolders for each genre (blues, pop, etc..)
AUDIO_ROOT_FOLDER = "Data/genres_original"
```

```python
# Even though the GTZAN already provides features (spectral properties) as a csv file.
# Lets build owr own, as we need this later on for predictions as well, where we will give the model
# a completely unrelated song and ask for finding recommendations from the GTZAN test.
# Also, we do not want skew the training and inference steps, where training is from a different feature set compared to inference.

# Function to process each file and create the spectral properties. 
def process_file_with_label(file_path, label):
    try:
        #Load audio in librosa 
        y, sr = librosa.load(file_path, duration=30) # GTZAN dataset is all 30-sec files.
    
        # extract the spectral features, Harmony, percussion and tempo
        chroma_stft = librosa.feature.chroma_stft(y=y, sr=sr)
        rms = librosa.feature.rms(y=y)
        spec_cent = librosa.feature.spectral_centroid(y=y, sr=sr)
        spec_bw = librosa.feature.spectral_bandwidth(y=y, sr=sr)
        rolloff = librosa.feature.spectral_rolloff(y=y, sr=sr)
        zcr = librosa.feature.zero_crossing_rate(y=y)
        harmony, perceptr = librosa.effects.hpss(y)
        tempo_raw, _ = librosa.beat.beat_track(y=y, sr=sr)

        if np.ndim(tempo_raw) > 0:
            # If it's an array (e.g. [120.5]), take the first item
            # Check if empty first to avoid index error
            if len(tempo_raw) > 0:
                tempo = float(tempo_raw[0])
            else:
                tempo = 0.0 # Fallback for empty arrays
        else:
            # If it's already a scalar float
            tempo = float(tempo_raw)
        
    
        # extract the MFCCs (Mel Frequency cepstral co-efficients)
        # 20 MFCCs are generally the standard in audio processing
        mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=20)
    except:
        # This is required. Found in data preparation that GTZAN has 1 corrupt file (in jazz genre).
        # Without this exception handling, the code will crash for the corrupt file.
        print(f"Exception in file {file_path}")
        return None

    # Create dictioary that will one row in the DataFrame
    row = {
        'file_path': os.path.basename(file_path),

        # Spectral means/vars
        'chroma_stft_mean': np.mean(chroma_stft), 'chroma_stft_var': np.var(chroma_stft),
        'rms_mean': np.mean(rms), 'rms_var': np.var(rms),
        'spectral_centroid_mean': np.mean(spec_cent), 'spectral_centroid_var': np.var(spec_cent),
        'spectral_bandwidth_mean': np.mean(spec_bw), 'spectral_bandwidth_var': np.var(spec_bw),
        'rolloff_mean': np.mean(rolloff), 'rolloff_var': np.var(rolloff),
        'zero_crossing_rate_mean': np.mean(zcr), 'zero_crossing_rate_var': np.var(zcr),
        'harmony_mean': np.mean(harmony), 'harmony_var': np.var(harmony),
        'perceptual_mean': np.mean(perceptr), 'perceptual_var': np.var(perceptr),
        'tempo': tempo,
        'label': label,
    }
    # Add MFCCs (mfcc1_mean, mfcc1_var ... mfcc20_var)
    for i in range(20):
        row[f'mfcc{i+1}_mean'] = np.mean(mfcc[i])
        row[f'mfcc{i+1}_var'] = np.var(mfcc[i])

    # Here we just return a dictionary with all the values
    return row
```

```python
# Helper function to get file_paths and genre_label for all files in a single genre
def get_all_file_tasks_per_genre(genre_dir_path):
    if not os.path.isdir(genre_dir_path):
        print(f"Provided path {genre_dir_path} is not a directory")
        return []

    genre_label = os.path.basename(genre_dir_path)
    file_tasks = []
    for file in os.listdir(genre_dir_path):
        if os.path.isdir(file):
            continue # ignore sub-directories in genre if any
        if not file.endswith('.wav'):
            continue # skip files that are not wav audio files 

        file_path = os.path.join(genre_dir_path, file)
        file_tasks.append((file_path, genre_label))

    return file_tasks
```

```python
# Helper function to get all the genre directories as a list
def get_genre_dirs():
    genre_dir_paths = []

    # Go through the top level folder and extract all directories into a list
    # The GTZAN dataset stores audio file per genre folder (ex: blues, classical, etc..) 
    for file in os.listdir(AUDIO_ROOT_FOLDER):
        file_path = os.path.join(AUDIO_ROOT_FOLDER, file)
        if os.path.isdir(file_path):
            genre_dir_paths.append(file_path)
        else:
            continue # if not directory, ignore the file.

    return genre_dir_paths
```

```python
# Helper function to extract the features from Audio (.wav) files 

# Note that this function first checks if computed features are already present, If so, it skips the feature extraction.
# After spending quite some time and adding parallelization, the feature extraction still takes quite some time
# on my Macbook with a M3 Pro CPU
def save_genre_features_from_audios(parallel, force_extraction):
    """
    parallel - if set to true, parallelizes feature extraction per genre
    force_extraction - Re-run (force) extraction even if computed_features.csv already exists
                        Note that even if this parameter is set to False, the feature extraction
                        will still happen if computed_features.csv does not exist.
    """
    

    # I have already provided the computed_features.csv in github along with this notebook. 
    # If you wish to re-run the extraction or update anything in the feature properties,
    # call this function with force_extraction = True
    if (not force_extraction) and os.path.exists("computed_features.csv"):
        print("Feature Extraction skipped")
        return

    print("Starting Feature Extraction")
    genre_dir_paths = get_genre_dirs()

    # I tried processing one file at a time and this is super slow on my M2 mac. 
    # Then attempted to parallelize all files, which was pretty fast, but crashing and not stable.
    # This version parallelizes files per genre. Seems a good midway point to achieve speed and stability 
    valid_results = []
    for genre_dir in genre_dir_paths:
        tasks = get_all_file_tasks_per_genre(genre_dir)
        # If there are no files in the genre folder, skip this folder.
        if len(tasks) == 0:
            continue

        results = []
        if parallel:
            print(f"Found {len(tasks)} for {genre_dir}. Starting parallel processing...")
            results = Parallel(n_jobs=-1, verbose=5)(
                delayed(process_file_with_label)(path, label) for path, label in tasks
            )
        else:
            print(f"Found {len(tasks)} for {genre_dir}. Starting sequential processing...")
            count = 1
            for path, label in tasks:
                count = count + 1
                results = process_file_with_label(path, label)
                if (count%10 == 0):
                    print(f"Done processing {count} files, continuing", end="", flush=True)

        for r in results:
            if r is not None:
                valid_results.append(r)
    
    df = pd.DataFrame(valid_results)
    df.to_csv("computed_features.csv", index=False)

    print("Finished creating computed_features.csv file.")
```

### Notes

We have successfully converted the audio files to a numerical representation with 57 features, 1 label and 1 file_path column

The rest of the notebook attempts to build a Baseline model using K-Nearest Neighbors. 
Choosing KNN as baseline, as for this dataset, we can imagine a 57 dimension hyperplane and given any song, we want to find other songs (points in the hyperplane) that are closest to the point for given song

```python
from sklearn.pipeline import Pipeline

# Scaler is needed as the range of the features is extremely uneven

pipeline_knn_with_scaler = Pipeline([
    ('standard_scaler', StandardScaler()),
    ('knn_model', NearestNeighbors(n_neighbors=5))
])
```

```python
# Drop the 'file_path' as that is a metadata column and drop the 'label' as that is the target column
df_features = df_computed_thirty.drop(columns=['file_path', 'label'])
df_features
```

```python
pipeline_knn_with_scaler.fit(df_features)
```

```python
import math

# function to get a Recommendation based on the k neighbors and distance to each 

# The Knn model is trained, but to get recommendations, we need to find the K (chosen 5 here for limiting the recommendations to 5)
# neighbors that are closest, given the song.

def get_neighbors_with_distance(song_index):
    """
    song_index - index of the song in the dataframe for which this function returns its neighbors.
    """
    song_features = df_features.iloc[[song_index]]

    scaled_song_features = pipeline_knn_with_scaler['standard_scaler'].transform(song_features)

    distances, indices = pipeline_knn_with_scaler.named_steps['knn_model'].kneighbors(scaled_song_features)

    song_filename = df_computed_thirty.iloc[song_index]['file_path']
    song_genre = df_computed_thirty.iloc[song_index]['label']
    print(f"Given Song :{song_filename} has genre {song_genre}")

    print(distances[0][1]) 
    neighbors = []
    print ("List of neighbors")
    for i in range(len(indices[0])):

        # we skip distance of 0, since knn also returns the distance to given input as well, which will always be 0
        # Found while testing that due to mathematical rounding the distance to itself is not always exactly 0, but
        # it is a very small number, so ignoraing anything with 1e-07
        if math.isclose(distances[0][i], 0.0, abs_tol=1e-07):
            continue

        neighbor_idx = indices[0][i]
        dist = distances[0][i]
        neighbor_filename = df_computed_thirty.iloc[neighbor_idx]['file_path']
        neighbor_genre = df_computed_thirty.iloc[neighbor_idx]['label']
        values_to_append =(neighbor_idx, dist, neighbor_filename, neighbor_genre)
        neighbors.append(values_to_append)
        print(f" Neighbor {i}, {neighbor_filename} with genre {neighbor_genre} and distance {dist}")

    return song_filename, song_genre, neighbors
```

```python
# Lets find the neighbors for the song at index 10

given_song_file_name, given_song_genre, neighbors = get_neighbors_with_distance(10)
```

```python
# Lets display the song and also lets play the neighbors (recommendations)

from IPython.display import Audio, display

print(f'Given song {given_song_file_name} has genre {given_song_genre}')
given_song_filepath = "Data/genres_original/" + given_song_genre + "/" + given_song_file_name
display(Audio(given_song_filepath))

for i in range(len(neighbors)):
    recommended_neighbor = neighbors[i]
    # Idx 3 is genre (which is also the folder path)
    filename_to_play = "Data/genres_original" + "/" + str(recommended_neighbor[3]) + "/" + str(recommended_neighbor[2]) 
    print(f'Neighbor priority {i+1}, {filename_to_play}, with distance {recommended_neighbor[1]}')
    display(Audio(filename_to_play))
```
